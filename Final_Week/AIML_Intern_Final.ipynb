```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder
```

```python
# Features (numeric only)
X = df[["Exposure", "Vulnerability", "Susceptibility",
              "Lack of Coping Capabilities", " Lack of Adaptive Capacities", "Year"]]

# Target (categorical: WRI Category)
y = df["WRI Category"]

# Encode target labels
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded)

# Random Forest Classifier
rf = RandomForestClassifier(random_state=42, n_estimators=100)
rf.fit(X_train, y_train)

# Predictions
y_pred = rf.predict(X_test)

# Evaluation
accuracy = accuracy_score(y_test, y_pred)
class_report = classification_report(y_test, y_pred, target_names=le.classes_)
conf_matrix = confusion_matrix(y_test, y_pred)

```

```python
print(f"Accuracy: {accuracy}")
```

```python
print("\nClassification Report:\n", class_report)
```

```python
print("\nConfusion Matrix:\n", conf_matrix)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
import pickle

# Load dataset
df = pd.read_csv("/content/cleaned_data.csv")

# Features & Target
X = df[
    [
        "Exposure",
        "Vulnerability",
        "Susceptibility",
        "Lack of Coping Capabilities",
        " Lack of Adaptive Capacities",
        "Year",
    ]
]
y = df["WRI Category"]

# Encode target labels
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

# Train model
rf = RandomForestClassifier(random_state=42, n_estimators=100)
rf.fit(X_train, y_train)

# Save model & encoder
with open("rf_model.pkl", "wb") as f:
    pickle.dump(rf, f)
with open("label_encoder.pkl", "wb") as f:
    pickle.dump(le, f)

print("✅ Model trained & saved.")
```

```python
import pickle
import numpy as np

# Load model & encoder
with open("rf_model.pkl", "rb") as f:
    rf = pickle.load(f)

with open("label_encoder.pkl", "rb") as f:
    le = pickle.load(f)

# Example input (manually enter values)
input_data = np.array([[29, 10, 12, 19, 20, 2010]])
# Format → [Exposure, Vulnerability, Susceptibility, Lack of Coping, Lack of Adaptive, Year]

# Prediction
pred_encoded = rf.predict(input_data)[0]
pred_label = le.inverse_transform([pred_encoded])[0]

print("🌍 Predicted WRI Category:", pred_label)
```
